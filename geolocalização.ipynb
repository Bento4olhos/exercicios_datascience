```python
import folium

# Definindo as coordenadas e nomes das cidades
cidades = [
    {"nome": "Porto Alegre, RS, Brasil", "latitude": -30.0346, "longitude": -51.2177},
    {"nome": "São Paulo, SP, Brasil", "latitude": -23.5505, "longitude": -46.6333},
    {"nome": "Rio de Janeiro, RJ, Brasil", "latitude": -22.9068, "longitude": -43.1729},
    {"nome": "Belo Horizonte, MG, Brasil", "latitude": -19.8157, "longitude": -43.9542},
    {"nome": "Curitiba, PR, Brasil", "latitude": -25.4294, "longitude": -49.2711}
]

# Calcular a média das coordenadas para centralizar o mapa
media_latitude = sum(cidade["latitude"] for cidade in cidades) / len(cidades)
media_longitude = sum(cidade["longitude"] for cidade in cidades) / len(cidades)

# Criar o mapa centrado na média das coordenadas
mapa = folium.Map(location=[media_latitude, media_longitude], zoom_start=6)

# Adicionar marcadores ao mapa
for cidade in cidades:
    folium.Marker(
        [cidade["latitude"], cidade["longitude"]],
        popup=cidade["nome"]
    ).add_to(mapa)

# Salvar o mapa em um arquivo HTML
mapa.save("mapa_cidades.html")
```
